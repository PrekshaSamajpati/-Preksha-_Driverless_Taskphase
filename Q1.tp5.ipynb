```python
import numpy as np

x, y = 10.0, 5.0
heading = np.radians(30)     # NumPy trig needs radians
speed = 5.0                  # m/s
accel = 1.0                  # m/s^2
yaw_rate = np.radians(10)    # rad/s

total_time = 0.2
dt = 0.0001
steps = int(total_time / dt)
```

```python
for i in range(steps):
    x += speed * np.cos(heading) * dt
    y += speed * np.sin(heading) * dt
    speed += accel * dt
    heading += yaw_rate * dt

print("New pose:", round(x, 3), round(y, 3), round(np.degrees(heading), 1))
```

```python
def to_car_frame(cones, car_x, car_y, car_heading):
    # Step 1: translate
    shifted = cones - np.array([car_x, car_y])

    # Step 2: rotate by minus the heading
    c, s = np.cos(car_heading), np.sin(car_heading)
    R = np.array([[ c, s],
                  [-s, c]])

    return (R @ shifted.T).T
```

```python
names = ["B1", "B2", "Y1", "Y2"]
cones = np.array([[12, 7],
                  [8, 6],
                  [13, 4],
                  [9, 3]])

start = to_car_frame(cones, 10, 5, np.radians(30))
new   = to_car_frame(cones, x, y, heading)

print("Cones at the START pose:")
for n, p in zip(names, start):
    print(n, np.round(p, 3))

print("\nCones at the NEW pose:")
for n, p in zip(names, new):
    print(n, np.round(p, 3))
```
